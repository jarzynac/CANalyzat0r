##Komórka 2: Inicjalizacja demona tła nieblokującego
Aby zapobiec zawieszaniu się środowiska notebooka, musimy utworzyć instancję usługi Ollama jako asynchronicznego podprocesu w tle przy użyciu Pythona:

```python
import os
import subprocess
import time

# Configure host binding and CORS allowance (CRITICAL for web clients)
os.environ['OLLAMA_HOST'] = '127.0.0.1:11434'
os.environ['OLLAMA_ORIGINS'] = '*'

# Launch the server using its absolute path to prevent FileNotFoundError
print("Initializing Ollama Daemon...")
subprocess.Popen(['/usr/local/bin/ollama', 'serve'])

# Provide a mandatory 10-second grace period for full initialization
time.sleep(10)
print("Daemon initialization complete. Port 11434 is now listening.")
```

##Komórka 3: Wchłanianie i weryfikacja modelu

```python
!pip install -q ollama

import subprocess
import ollama

# Pull the optimized roleplay model slug (using a valid model tag)
print("Downloading qwen2.5:7b...")
subprocess.run(["/usr/local/bin/ollama", "pull", "qwen2.5:7b"], check=True)

# Run a quick generation test directly from the Python SDK to confirm execution
print("Verifying model generation...")
response = ollama.chat(model='qwen2.5:7b', messages=[
    {'role': 'user', 'content': 'Hello! Respond with a single short sentence.'}
])
print("\nTest Output:\n", response['message']['content'])
```
